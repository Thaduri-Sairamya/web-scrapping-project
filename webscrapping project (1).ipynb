```python

import requests
from bs4 import BeautifulSoup
import re
```

problem statement:

To analyze Royal Enfield bike models based on price, mileage, engine capacity, power, and weight, and extract insights that can help customers and businesses make better decisions

```python

url="https://www.bikewale.com/royalenfield-bikes/"
```

```python
url
```

```python
headers={
    "user-agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/149.0.0.0 Safari/537.36",
    
}
```

```python
response=requests.get(url,headers=headers)
response.status_code
```

```python
html_data=response.text
soup=BeautifulSoup(html_data)
```

```python
soup.text
```

```python
#DATA COLLECTION
product_name=[]
name=soup.find_all("h3",class_="o-j4 o-jq o-hM o-c4 o-jK")
for i in name:
    product_name.append(i.text)
```

```python
product_name
```

```python
product_price=[]
price=soup.find_all("span",class_="o-jJ o-jr o-j3 o-kJ")
for i in price:
    product_price.append(i.text)
```

```python
product_price
```

```python
product_rating=[]
cc=soup.find_all("span",class_="o-jL o-f7 o-j0 o-jc o-fE")
for i in cc:
    product_rating.append(i.text)
```

```python
product_rating
```

```python
product_mileage = []
mileage_tags = soup.find_all("span", class_="o-jK")
for m in mileage_tags:
    text = m.text.strip()
    if "kmpl" in text.lower():
        product_mileage.append(text)
```

```python
product_mileage
```

```python
product_capacity=[]
cc=soup.find_all("span",class_="o-jK")
for i in cc:
    text=i.text.strip()
    if "cc" in text.lower():
        product_capacity.append(text)
```

```python
product_capacity
```

```python
product_power=[]
power=soup.find_all("span",class_="o-jK")
for i in power:
    text=i.text.strip()
    if "bhp" in text.lower():
      product_power.append(text)
```

```python
product_power
```

```python
product_weight=[]
weight=soup.find_all("span",class_="o-jK")
for i in weight:
    text=i.text.strip()
    if "kg" in text.lower():
      product_weight.append(text)
```

```python
product_weight
```

```python
min_len=min(len(product_name),
            len(product_price),
            len(product_rating),
            len(product_mileage),
            len(product_capacity),
            len(product_weight),
            len(product_power)
           )
            
```

```python
import numpy as np
import pandas as pd
df=pd.DataFrame({"product_name":product_name[:min_len],
                 "product_price":product_price[:min_len],
                 "product_rating":product_rating[:min_len],
                 "product_mileage":product_mileage[:min_len],
                 "product_capacity":product_capacity[:min_len],
                 "product_power":product_power[:min_len],
                 "product_weight":product_weight[:min_len]})
```
